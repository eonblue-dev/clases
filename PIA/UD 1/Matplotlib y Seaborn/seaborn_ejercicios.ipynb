## Ejercicio 3

Muestra un gráfico de líneas con tres líneas de distintos colores y grosores. Cada una de ellas representará los valores de las funciones trigonométricas seno, coseno y tangente entre los 0 y los 720 grados. Además, muestra un título y las etiquetas (labels):

```python
# grados de 0 a 720
grados = np.linspace(0, 720, 500)

# numpy trabaja en radianes así que hay que pasar los grados a radianes
radianes = np.radians(grados)

seno = np.sin(radianes)
coseno = np.cos(radianes)
tangente = np.tan(radianes)

sns.lineplot(x=grados, y=seno, label="seno", color="blue", linewidth=1)
sns.lineplot(x=grados, y=coseno, label="coseno", color="red", linewidth=3)
sns.lineplot(x=grados, y=tangente, label="tangente", color="green", linewidth=2)

# la tangente se va al infinito en 90, 270... y no se ven las otras líneas,
# así que limito el eje Y
plt.ylim(-3, 3)

plt.title("Funciones trigonométricas")
plt.xlabel("Grados")
plt.ylabel("Valor")
plt.legend()
plt.grid()

plt.show()
```

## Ejercicio 4

Muestra un gráfico de barras comparando dos conjuntos de datos elegidos por tí:

```python
# notas de dos alumnos en 5 asignaturas
asignaturas = ["Matemáticas", "Lengua", "Historia", "Inglés", "Ciencias"]
alumno1 = [7,5,9,6,8]
alumno2 = [6,8,7,9,5]

# junto los datos en un DataFrame para poder separar los alumnos con hue
datos = pd.DataFrame({
    "Asignatura": asignaturas * 2,
    "Nota": alumno1 + alumno2,
    "Alumno": ["Alumno 1"] * 5 + ["Alumno 2"] * 5
})

sns.barplot(data=datos, x="Asignatura", y="Nota", hue="Alumno",
            palette=["lightblue", "orange"], errorbar=None)

plt.title("Notas de dos alumnos")
plt.xlabel("Asignatura")
plt.ylabel("Nota")
plt.legend()

plt.show()
```

## Ejercicio 7

Muestra en un gráfico de dispersión dos conjuntos de datos aleatorios generados por tí. Estos datos deben ser valores comprendidos entre el 0 y el 25, tanto en el eje X como en el eje Y. Cada conjunto se pintará de un color distinto:

```python
# dos conjuntos de datos aleatorios entre 0 y 25
# pongo 26 porque randint no incluye el número de arriba
x1 = np.random.randint(26, size=[20])
y1 = np.random.randint(26, size=[20])
x2 = np.random.randint(26, size=[20])
y2 = np.random.randint(26, size=[20])

# junto los dos conjuntos en un DataFrame
datos1 = pd.DataFrame({"X": x1, "Y": y1, "Conjunto": "Datos 1"})
datos2 = pd.DataFrame({"X": x2, "Y": y2, "Conjunto": "Datos 2"})
datos = pd.concat([datos1, datos2])

sns.scatterplot(data=datos, x="X", y="Y", hue="Conjunto",
                palette=["blue", "red"])

plt.title("Gráfico de dispersión")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.legend()

plt.show()
```

Muestra un gráfico circular estilizado con la distribución de tus horas del día:

```python
# mis horas del día (suman 24)
horas = [8,6,2,3,2,3]
actividades = ["Dormir", "Clase", "Comer", "Estudiar", "Deporte", "Ocio"]
colores = sns.color_palette("pastel", 6)

# Seaborn no tiene una función para gráficos circulares, por eso uso plt.pie
plt.pie(horas, labels=actividades, colors=colores, startangle=90,
        shadow=True, explode=(0,0,0,0.2,0,0), autopct="%1.1f%%")

plt.title("Distribución de mis horas del día")
plt.show()
```

## Ejercicio 10

Hay una página web llamada [Kaggle](https://www.kaggle.com/datasets) donde se publican distintos datasets y trabajos de machine learning sobre los mismos. Descárgate un dataset de tu elección y muestra datos relevantes del mismo utilizando las gráficas vistas hasta ahora:

```python
# no he podido bajar uno de Kaggle, así que uso el csv de covid que hay en la carpeta
# son datos acumulados de España de 2020. Lo leo con pandas que ya lo vimos.
df = pd.read_csv("03-nacional_covid19.csv")
print(df.head())
print(df.shape)

# hay columnas con muchos nulos
print(df.isnull().sum())

# --- Gráfica 1: líneas con los casos PCR y los fallecimientos acumulados ---
sns.lineplot(data=df, x="fecha", y="casos_pcr", label="Casos PCR", color="blue")
sns.lineplot(data=df, x="fecha", y="fallecimientos", label="Fallecimientos", color="red")

plt.title("COVID-19 en España (acumulado)")
plt.xlabel("Fecha")
plt.ylabel("Personas")
plt.legend()
plt.grid()

# hay muchas fechas y se pisan, así que enseño solo una de cada 30
plt.xticks(df["fecha"][::30], rotation=45)
plt.show()

# --- Gráfica 2: barras con los fallecimientos acumulados al final de cada mes ---
# saco el mes de la fecha (los 7 primeros caracteres, ej: 2020-03)
df["mes"] = df["fecha"].str[:7]
fallecidos_mes = df.groupby("mes")["fallecimientos"].max().reset_index()

sns.barplot(data=fallecidos_mes, x="mes", y="fallecimientos",
            color="orange", errorbar=None)

plt.title("Fallecimientos acumulados por mes")
plt.xlabel("Mes")
plt.ylabel("Fallecimientos")
plt.xticks(rotation=45)
plt.show()

# --- Gráfica 3: dispersión hospitalizados vs ingresos en UCI ---
sns.scatterplot(data=df, x="hospitalizados", y="ingresos_uci", color="purple")

plt.title("Hospitalizados vs ingresos en UCI")
plt.xlabel("Hospitalizados (acumulado)")
plt.ylabel("Ingresos UCI (acumulado)")
plt.show()
```

## Ejercicio 11

Sobre el mismo dataset anterior u otro nuevo, muestra tres gráficas distintas a las aprendidas hasta el momento. Para ello puedes consultar la web de la [documentación de Matplotlib](https://matplotlib.org/stable/gallery/index.html):

```python
# uso el mismo dataset de covid. He mirado en la documentación
# y he escogido: boxplot, fill_between y barras horizontales

# --- Gráfica 1: diagrama de cajas (boxplot) ---
# quito los nulos
hospitalizados = df["hospitalizados"].dropna()
uci = df["ingresos_uci"].dropna()

datos = pd.DataFrame({
    "Personas": pd.concat([hospitalizados, uci], ignore_index=True),
    "Tipo": ["Hospitalizados"] * len(hospitalizados) + ["UCI"] * len(uci)
})

sns.boxplot(data=datos, x="Tipo", y="Personas")
plt.title("Diagrama de cajas")
plt.ylabel("Personas (acumulado)")
plt.show()

# --- Gráfica 2: área rellena ---
# uso un número para el eje X (el día)
dias = np.arange(len(df))
fallecidos = df["fallecimientos"].fillna(0)

# uso Seaborn para dibujar la línea y plt para rellenar el área
sns.lineplot(x=dias, y=fallecidos, color="red")
plt.fill_between(dias, fallecidos, color="red", alpha=0.4)

plt.title("Fallecimientos acumulados")
plt.xlabel("Día (desde el primer registro)")
plt.ylabel("Fallecimientos")
plt.show()

# --- Gráfica 3: barras horizontales ---
# saco el valor máximo de cada columna
columnas = ["casos_pcr", "hospitalizados", "altas", "fallecimientos", "ingresos_uci"]
maximos = []

for col in columnas:
    maximos.append(df[col].max())

datos_maximos = pd.DataFrame({
    "Indicador": columnas,
    "Máximo": maximos
})

sns.barplot(data=datos_maximos, x="Máximo", y="Indicador", color="lightblue",
            errorbar=None)

plt.title("Valor máximo de cada columna")
plt.xlabel("Personas")
plt.show()
```
