## Ejercicio 3

Muestra un gráfico de líneas con tres líneas de distintos colores y grosores. Cada una de ellas representará los valores de las funciones trigonométricas seno, coseno y tangente entre los 0 y los 720 grados. Además, muestra un título y las etiquetas (labels):

```python
# grados de 0 a 720
grados = np.linspace(0, 720, 500)

# numpy trabaja en radianes así que hay que pasar los grados a radianes
radianes = np.radians(grados)

seno = np.sin(radianes)
coseno = np.cos(radianes)
tangente = np.tan(radianes)

plt.plot(grados, seno, label = "seno", color = "blue", linewidth = 1)
plt.plot(grados, coseno, label = "coseno", color = "red", linewidth = 3)
plt.plot(grados, tangente, label = "tangente", color = "green", linewidth = 2)

# la tangente se va al infinito en 90, 270... y no se ven las otras líneas,
# así que limito el eje Y (esto lo he buscado, no venía en la teoría)
plt.ylim(-3, 3)

plt.title("Funciones trigonométricas")
plt.xlabel("Grados")
plt.ylabel("Valor")
plt.legend()
plt.grid()

plt.show()
```

## Ejercicio 10

Hay una página web llamada [Kaggle](https://www.kaggle.com/datasets) donde se publican distintos datasets y trabajos de machine learning sobre los mismos. Descárgate un dataset de tu elección y muestra datos relevantes del mismo utilizando las gráficas vistas hasta ahora:

```python
# no he podido bajar uno de Kaggle, así que uso el csv de covid que hay en la carpeta
# (datos acumulados de España de 2020). Lo leo con pandas que ya lo vimos.
import pandas as pd

df = pd.read_csv("03-nacional_covid19.csv")
print(df.head())
print(df.shape)

# hay columnas con muchos nulos
print(df.isnull().sum())

# --- Gráfica 1: líneas con los casos PCR y los fallecimientos acumulados ---
plt.plot(df["fecha"], df["casos_pcr"], label = "Casos PCR", color = "blue")
plt.plot(df["fecha"], df["fallecimientos"], label = "Fallecimientos", color = "red")

plt.title("COVID-19 en España (acumulado)")
plt.xlabel("Fecha")
plt.ylabel("Personas")
plt.legend()
plt.grid()

# hay muchas fechas y se pisan, así que enseño solo una de cada 30
plt.xticks(df["fecha"][::30], rotation = 45)
plt.show()

# --- Gráfica 2: barras con los fallecimientos acumulados al final de cada mes ---
# saco el mes de la fecha (los 7 primeros caracteres, ej: 2020-03)
df["mes"] = df["fecha"].str[:7]
fallecidos_mes = df.groupby("mes")["fallecimientos"].max()

plt.bar(fallecidos_mes.index, fallecidos_mes.values, color = "orange")
plt.title("Fallecimientos acumulados por mes")
plt.xlabel("Mes")
plt.ylabel("Fallecimientos")
plt.xticks(rotation = 45)
plt.show()

# --- Gráfica 3: dispersión hospitalizados vs ingresos en UCI ---
plt.scatter(df["hospitalizados"], df["ingresos_uci"], color = "purple")
plt.title("Hospitalizados vs ingresos en UCI")
plt.xlabel("Hospitalizados (acumulado)")
plt.ylabel("Ingresos UCI (acumulado)")
plt.show()
```

## Ejercicio 11

Sobre el mismo dataset anterior u otro nuevo, muestra tres gráficas distintas a las aprendidas hasta el momento. Para ello puedes consultar la web de la [documentación de Matplotlib](https://matplotlib.org/stable/gallery/index.html):

```python
# uso el mismo dataset de covid. He mirado en la documentación de matplotlib
# y he escogido: boxplot, fill_between y barh

# --- Gráfica 1: diagrama de cajas (boxplot) ---
# quito los nulos porque si no boxplot no funciona bien
hospitalizados = df["hospitalizados"].dropna()
uci = df["ingresos_uci"].dropna()

plt.boxplot([hospitalizados, uci], tick_labels = ["Hospitalizados", "UCI"])
plt.title("Diagrama de cajas")
plt.ylabel("Personas (acumulado)")
plt.show()

# --- Gráfica 2: área rellena (fill_between) ---
# uso un número para el eje X (el día) porque con las fechas en texto da problemas
dias = np.arange(len(df))
fallecidos = df["fallecimientos"].fillna(0)

plt.fill_between(dias, fallecidos, color = "red", alpha = 0.4)
plt.plot(dias, fallecidos, color = "red")
plt.title("Fallecimientos acumulados")
plt.xlabel("Día (desde el primer registro)")
plt.ylabel("Fallecimientos")
plt.show()

# --- Gráfica 3: barras horizontales (barh) ---
# el valor máximo de cada columna (como son acumulados es el último dato que hay)
columnas = ["casos_pcr", "hospitalizados", "altas", "fallecimientos", "ingresos_uci"]
maximos = []
for col in columnas:
    maximos.append(df[col].max())

plt.barh(columnas, maximos, color = "lightblue")
plt.title("Valor máximo de cada columna")
plt.xlabel("Personas")
plt.show()
```
