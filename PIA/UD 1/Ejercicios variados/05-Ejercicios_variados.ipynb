## Ejercicio 2

Crea una matriz de NumPy de dos dimensiones (10 filas x 2 columnas) a partir de una lista de listas de Python. Muestra un gráfico de líneas teniendo en cuenta que el índice de la fila es el eje X (las posiciones 0,2,3...), mientras que los datos de cada columna serán los valores del eje Y. Como hay dos columnas deberás dibujar dos líneas en el gráfico. Por último, el gráfico debe añadir lo siguiente:

*   Deberá mostrar las etiquetas de las líneas
*   Deberá tener el título "Gráfico de líneas"
*   Las líneas deberán estar dibujadas de distintos colores
*   Cada línea deberá tener un grosor distinto
*   Cada eje deberá tener su propio título

```python
# lista de listas: 10 filas y 2 columnas
lista = [[1, 10],
         [2, 9],
         [4, 8],
         [3, 7],
         [5, 7],
         [7, 5],
         [6, 4],
         [8, 3],
         [9, 2],
         [10, 1]]

matriz = np.array(lista)
print(matriz.shape)

# el eje X es la posición de la fila (0, 1, 2... 9)
x = np.arange(len(matriz))

# cada columna es una línea
plt.plot(x, matriz[:, 0], label = "Columna 0", color = "blue", linewidth = 2)
plt.plot(x, matriz[:, 1], label = "Columna 1", color = "red", linewidth = 5)

plt.title("Gráfico de líneas")
plt.xlabel("Índice de la fila")
plt.ylabel("Valor")
plt.legend()
plt.grid()
plt.show()
```

## Ejercicio 3

Genera una matriz (100 filas x 5 columnas) de números enteros aleatorios del 0 al 9 con NumPy. Muestra la siguiente información de la misma:

*   Tamaño (cantidad de elementos en total)
*   Forma
*   Tipo (recuerda utilizar SOLO NumPy)
*   Desviación estándar
*   Media
*   Mediana
*   Todos los valores entre las filas 25 y 27 (ambas incluídas) y entre las columnas 1 y 3
*   Matriz de correlación entre columnas

```python
# 100 filas x 5 columnas con números del 0 al 9 (randint no incluye el 10)
matriz = np.random.randint(10, size = (100, 5))

print("Tamaño:", matriz.size)
print("Forma:", matriz.shape)
print("Tipo:", matriz.dtype)
print("Desviación estándar:", np.std(matriz))
print("Media:", np.mean(matriz))
print("Mediana:", np.median(matriz))

# filas 25 a 27 incluidas -> 25:28, columnas 1 a 3 incluidas -> 1:4
print("Filas 25-27 y columnas 1-3:")
print(matriz[25:28, 1:4])

# np.corrcoef usa las filas como variables, así que le pongo rowvar = False
# para que use las columnas
print("Matriz de correlación entre columnas:")
print(np.corrcoef(matriz, rowvar = False))
```

Genera una matriz (50 filas x 6 columnas) de datos aleatorios entre 0 y 1 con NumPy. Muestra un gráfico de dispersión teniendo en cuenta que los datos contenidos en cada par de columnas pertenecerán a clases distintas (dibujar cada par de columnas con un color distinto). Por ejemplo:

## Ejercicio 6

Sobre el dataset "iris", calcula la media de cada atributo para cada variedad de la flor. Después muestra un gráfico de barras que compare dichos valores:

```python
# media de cada columna para cada variedad
medias = iris.groupby("variety").mean()
print(medias)

# 4 atributos, así que pongo 4 barras por variedad desplazadas (como en la teoría)
x = np.arange(len(medias))   # 0, 1, 2 -> una posición por variedad
ancho = 0.2

plt.bar(x - 0.3, medias["sepal.length"], width = ancho, label = "sepal.length", color = "lightblue")
plt.bar(x - 0.1, medias["sepal.width"], width = ancho, label = "sepal.width", color = "orange")
plt.bar(x + 0.1, medias["petal.length"], width = ancho, label = "petal.length", color = "green")
plt.bar(x + 0.3, medias["petal.width"], width = ancho, label = "petal.width", color = "red")

# pongo el nombre de la variedad debajo de cada grupo de barras
plt.xticks(x, medias.index)

plt.title("Media de cada atributo por variedad")
plt.xlabel("Variedad")
plt.ylabel("Media (cm)")
plt.legend()
plt.show()
```

## Ejercicio 10

Partimos del dataset "mount_everest_deaths.csv" que contiene la siguiente información por columnas:
*   **No.**: columna con un número identificativo para cada línea.
*   **Name**: Nombre del fallecido
*   **Date**: Fecha del accidente
*   **Age**: Edad del fallecido
*   **Expedition**: Nombre de la expedición
*   **Nationality**: Nacionalidad del fallecido
*   **Cause of death**: Causa de la muerte
*   **Location**: Lugar de la muerte

Muestra una gráfica que consideres relevante. Para ello, deja en el código con comentarios todo el trabajo exploratorio que hagas sobre el dataset:

```python
everest = pd.read_csv("mount_everest_deaths.csv")

# ---- EXPLORACIÓN ----
# miro las primeras filas y el tamaño
print(everest.head())
print(everest.shape)   # 309 filas y 8 columnas

# tipos de cada columna: Date es texto, no fecha
print(everest.dtypes)

# nulos: Age tiene casi la mitad vacía (150), así que no me fío mucho de la edad
print(everest.isnull().sum())

# causas de muerte más frecuentes
# OJO: hay causas que son lo mismo escritas distinto (ej: "Avalanche" y
# "2014 Mount Everest Avalanche"), así que contar causas es un poco lioso
print(everest["Cause of death"].value_counts().head(10))

# nacionalidades: la mayoría son de Nepal (sherpas)
print(everest["Nationality"].value_counts().head(10))

# las fechas son tipo "June 7, 1922", así que el año son los 4 últimos caracteres
print(everest["Date"].str[-4:].unique())
# sale un valor raro "g-75" -> hay una fila con la fecha "Aug-75"
print(everest[everest["Date"].str[-4:] == "g-75"])

# ---- PREPARO LOS DATOS ----
everest["Year"] = everest["Date"].str[-4:]
# arreglo a mano la fila de "Aug-75" (es de la expedición de 1975)
everest.loc[everest["Date"] == "Aug-75", "Year"] = "1975"
everest["Year"] = everest["Year"].astype(int)

# agrupo por décadas para que se vea mejor que año a año
everest["Decade"] = (everest["Year"] // 10) * 10
por_decada = everest["Decade"].value_counts().sort_index()
print(por_decada)

# ---- GRÁFICA ----
# me parece relevante ver cómo han cambiado las muertes con el tiempo
plt.bar(por_decada.index, por_decada.values, width = 8, color = "steelblue")
plt.title("Muertes en el Everest por década")
plt.xlabel("Década")
plt.ylabel("Número de muertes")
plt.xticks(por_decada.index, rotation = 45)
plt.show()

# CONCLUSIÓN: las muertes suben mucho a partir de los 70-80, seguramente porque
# empieza a subir mucha más gente (no tengo el dato de cuánta gente sube,
# así que no puedo saber si es más peligroso o solo que hay más escaladores).
# La década de 2010 es la peor, pero hay que tener en cuenta las avalanchas de
# 2014 y la del terremoto de 2015, que mataron a mucha gente de golpe.
# La barra de 2020 es pequeña porque el dataset solo llega a 2021, la década no está completa.
```
