# Variante 01 — Ventas de tienda online

**Variable principal:** `importe` · **DataFrame:** `df_pedidos` · **Agrupación:** `categoria`

En esta variante se conservan los nombres de las columnas que aparecen en el enunciado.

````python
## Ejercicio 1 — NumPy: estadísticas de importe

datos = np.array([45, 80, 23, 120, 65, 32, 150, 90, 55, 110, 70, 40], dtype=float)

media = np.mean(datos)
mediana = np.median(datos)
valor_maximo = np.max(datos)
valor_minimo = np.min(datos)
desviacion_tipica = np.std(datos)
por_encima_media = datos[datos > media]

print("Datos:", datos)
print(f"Media: {media:.2f} €")
print(f"Mediana: {mediana:.2f} €")
print(f"Máximo: {valor_maximo:.2f} €")
print(f"Mínimo: {valor_minimo:.2f} €")
print(f"Desviación típica: {desviacion_tipica:.2f} €")
print("Valores por encima de la media:", por_encima_media)
````

````python
## Ejercicio 2 — Pandas: creación e inspección del DataFrame

# Los datos de esta tabla son ficticios y tienen 12 registros distintos.
df_pedidos = pd.DataFrame({
    "id": range(1, 13),
    "categoria": ['Electrónica', 'Ropa', 'Hogar', 'Electrónica', 'Ropa', 'Hogar', 'Electrónica', 'Ropa', 'Hogar', 'Electrónica', 'Ropa', 'Hogar'],
    "importe": datos,
    "ciudad": ['Linares', 'Granada', 'Jaén', 'Málaga', 'Linares', 'Granada', 'Jaén', 'Málaga', 'Linares', 'Granada', 'Jaén', 'Málaga'],
    "fecha": pd.date_range("2025-01-01", periods=12, freq="D")
})

# Introducimos un nulo en una columna auxiliar para practicar isna() y dropna().
df_pedidos.loc[11, "ciudad"] = np.nan

# Añadimos una fila duplicada para poder demostrar drop_duplicates() en el ejercicio 4.
df_pedidos = pd.concat([df_pedidos, df_pedidos.iloc[[0]]], ignore_index=True)

print("Primeras filas:")
print(df_pedidos.head())

print("\nTipos de datos:")
print(df_pedidos.dtypes)

print("\nValores nulos por columna:")
print(df_pedidos.isna().sum())

print("\nNúmero de filas duplicadas:", df_pedidos.duplicated().sum())

umbral = 60
print(f"\nRegistros con importe superior a {umbral}:")
print(df_pedidos[df_pedidos["importe"] > umbral])

print("\nPromedio de importe agrupado por categoria:")
print(df_pedidos.groupby("categoria")["importe"].mean().round(2))
````

````python
## Ejercicio 3 — Visualizaciones con Matplotlib y Seaborn

# Matplotlib: comparamos el valor de cada registro.
df_grafica = df_pedidos.sort_values("id")
plt.figure(figsize=(9, 4))
plt.plot(df_grafica["id"], df_grafica["importe"], marker="o")
plt.title("Comparación de importe por registro — Variante 01")
plt.xlabel("ID del registro")
plt.ylabel("importe (€)")
plt.tight_layout()
plt.show()

# Seaborn: comparamos la media de la variable principal por grupo.
plt.figure(figsize=(8, 4))
sns.barplot(data=df_pedidos, x="categoria", y="importe", estimator=np.mean, errorbar=None, color="C0")
plt.title("Media de importe por categoria")
plt.xlabel("categoria")
plt.ylabel("Media de importe (€)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

medias_grupo = df_pedidos.groupby("categoria")["importe"].mean().sort_values(ascending=False)
print("\nInterpretación:")
print(f"El grupo con mayor media es {medias_grupo.index[0]}, con {medias_grupo.iloc[0]:.2f} €.")
print(f"El grupo con menor media es {medias_grupo.index[-1]}, con {medias_grupo.iloc[-1]:.2f} €.")
````

````python
## Ejercicio 4 — Limpieza e informe completo

# 1. Eliminamos filas con algún dato nulo y registros duplicados.
filas_antes = len(df_pedidos)
df_limpio = df_pedidos.dropna().drop_duplicates().copy()
print(f"Filas antes de limpiar: {filas_antes}")
print(f"Filas después de limpiar: {len(df_limpio)}")

# 2. Creamos una columna calculada a partir de la variable principal.
#    Indica qué porcentaje representa cada registro respecto al valor máximo.
df_limpio["porcentaje_maximo"] = (
    df_limpio["importe"] / df_limpio["importe"].max() * 100
).round(1)

# 3. Ordenamos los registros de mayor a menor valor.
df_ordenado = df_limpio.sort_values("importe", ascending=False)
print("\nRegistros ordenados de mayor a menor:")
print(df_ordenado[["id", "categoria", "importe", "porcentaje_maximo"]])

# 4. Tabla resumen agrupada.
resumen = df_limpio.groupby("categoria").agg(
    media=("importe", "mean"),
    mediana=("importe", "median"),
    registros=("importe", "count"),
    porcentaje_medio_maximo=("porcentaje_maximo", "mean")
).sort_values("media", ascending=False).round(2)

print("\nTabla resumen:")
print(resumen)

# 5. Visualización final con Seaborn.
plt.figure(figsize=(8, 4))
sns.barplot(data=resumen.reset_index(), x="categoria", y="media", estimator=np.mean, errorbar=None, color="C0")
plt.title("Media de importe por categoria (datos limpios)")
plt.xlabel("categoria")
plt.ylabel("Media de importe (€)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

# Conclusiones automáticas: cinco líneas que se pueden incluir en el pequeño informe.
grupo_mayor = resumen["media"].idxmax()
grupo_menor = resumen["media"].idxmin()
registro_maximo = df_ordenado.iloc[0]
registro_minimo = df_ordenado.iloc[-1]
print("\nConclusiones:")
print(f"1. Después de eliminar nulos y duplicados, quedan {len(df_limpio)} registros válidos.")
print(f"2. La media general de importe es {df_limpio['importe'].mean():.2f} €.")
print(f"3. El grupo con mayor media es {grupo_mayor} ({resumen.loc[grupo_mayor, 'media']:.2f} €).")
print(f"4. El grupo con menor media es {grupo_menor} ({resumen.loc[grupo_menor, 'media']:.2f} €).")
print(f"5. El valor individual más alto es {registro_maximo['importe']} y el más bajo es {registro_minimo['importe']} €.")
````

## Bloque A — Variantes 1, 5, 9, 13 y 17

### 1. Filtro booleano de NumPy

```python
a = np.array([2, 4, 6, 8])
print(a[a > 5] * 2)
```

Primero `a > 5` crea la máscara `[False, False, True, True]`. La selección devuelve `[6, 8]` y al multiplicar por 2 se obtiene **`[12 16]`**.

### 2. `shape`, `reshape` y slicing de columnas

`np.arange(12)` genera los números del 0 al 11. `reshape(3, 4)` los organiza en 3 filas y 4 columnas, por lo que `m.shape` devuelve **`(3, 4)`**. La expresión `m[:, 1:3]` selecciona todas las filas y las columnas de índice 1 y 2:

```text
[[ 1  2]
 [ 5  6]
 [ 9 10]]
```

La matriz original tiene forma 3 × 4; la selección tiene forma 3 × 2.

### 3. Selección de columna y filtro

```python
df = pd.DataFrame({'grupo': ['A', 'A', 'B'], 'valor': [5, 8, 3]})
print(df['valor'].mean())
print(df[df['grupo'] == 'A']['valor'].sum())
```

- El primer `print` calcula la media de toda la columna `valor`: `(5 + 8 + 3) / 3 = **5.3333...**`.
- El segundo filtra las filas del grupo A (valores 5 y 8) y suma esos valores: **13**.
- `df['valor']` selecciona una columna; `df[df['grupo'] == 'A']` selecciona filas que cumplen una condición.

### 4. `sns.barplot`

` sns.barplot(data=df, x='grupo', y='valor')` crea un diagrama de barras. `data` indica el DataFrame, `x` la variable categórica del eje horizontal y `y` la variable numérica del eje vertical. Por defecto, cada barra representa la media de `valor` para cada grupo: A = 6.5 y B = 3.0. Seaborn puede mostrar intervalos de confianza alrededor de las medias.

````python
# Comprobación práctica del bloque A
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

a = np.array([2, 4, 6, 8])
print("1)", a[a > 5] * 2)

m = np.arange(12).reshape(3, 4)
print("2) shape:", m.shape)
print(m[:, 1:3])

df_conceptos_a = pd.DataFrame({"grupo": ["A", "A", "B"], "valor": [5, 8, 3]})
print("3) Media:", df_conceptos_a["valor"].mean())
print("3) Suma del grupo A:", df_conceptos_a[df_conceptos_a["grupo"] == "A"]["valor"].sum())

plt.figure(figsize=(6, 4))
sns.barplot(data=df_conceptos_a, x="grupo", y="valor", errorbar=None, color="C0")
plt.title("Media del valor por grupo")
plt.xlabel("Grupo")
plt.ylabel("Valor medio")
plt.tight_layout()
plt.show()
````

## Bloque B — Variantes 2, 6, 10, 14 y 18

### 1. Broadcasting de NumPy

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
b = a + np.array([10, 20, 30])
```

El array de una dimensión `[10, 20, 30]` se aplica a cada fila de la matriz. NumPy permite esta operación porque las dimensiones son compatibles (*broadcasting*). El resultado es:

```text
[[11 22 33]
 [14 25 36]]
```

### 2. Detección y sustitución de nulos

Para `df = DataFrame({'x':[1, None, 3], 'y':[4, 5, None]})`, `df.isna().sum()` devuelve un nulo en cada columna:

```text
x    1
y    1
```

`df.fillna(0)` devuelve una tabla nueva con los valores ausentes sustituidos por 0:

```text
     x    y
0  1.0  4.0
1  0.0  5.0
2  3.0  0.0
```

Por defecto, `fillna` no modifica el DataFrame original: habría que asignar el resultado (`df = df.fillna(0)`) o usar `inplace=True`.

### 3. `groupby` y media

```python
df.groupby('tipo')['precio'].mean()
```

Agrupa las filas por `tipo` y calcula la media de `precio` en cada grupo: **A = 20** (media de 10 y 30) y **B = 30** (media de 20 y 40).

### 4. `plt.figure` y `plt.plot`

`plt.figure(figsize=(6, 4))` crea una figura de 6 pulgadas de ancho por 4 de alto. `plt.plot([1,2,3], [4,1,5], marker='o')` dibuja una línea que une los puntos (1,4), (2,1) y (3,5); `marker='o'` añade un marcador circular a cada punto. `plt.show()` muestra la gráfica.

````python
# Comprobación práctica del bloque B
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

a = np.array([[1, 2, 3], [4, 5, 6]])
b = a + np.array([10, 20, 30])
print("1) Broadcasting:\n", b)

df_nulos = pd.DataFrame({"x": [1, None, 3], "y": [4, 5, None]})
print("2) Número de nulos:\n", df_nulos.isna().sum())
print("2) Sustituir nulos por cero:\n", df_nulos.fillna(0))

df_grupos = pd.DataFrame({"tipo": ["A", "B", "A", "B"], "precio": [10, 20, 30, 40]})
print("3) Media por tipo:\n", df_grupos.groupby("tipo")["precio"].mean())

plt.figure(figsize=(6, 4))
plt.plot([1, 2, 3], [4, 1, 5], marker="o")
plt.title("Ejemplo de plt.plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.tight_layout()
plt.show()
````

## Bloque C — Variantes 3, 7, 11, 15 y 19

### 1. Diferencia entre slicing y copia

```python
a = np.array([10, 20, 30, 40])
b = a[1:3]
b[0] = 99
print(a)
```

En NumPy, un slicing básico como `a[1:3]` normalmente crea una **vista** del array original, no una copia independiente. Por eso, cambiar `b[0]` también modifica el elemento correspondiente de `a`. La salida es **`[10 99 30 40]`**. Para evitarlo se puede usar `b = a[1:3].copy()`.

### 2. Filtrado de filas

```python
df = pd.DataFrame({'nombre': ['Ana', 'Luis', 'Eva'], 'nota': [6, 4, 9]})
print(df[df['nota'] >= 5])
```

La condición conserva las filas cuya nota es mayor o igual que 5. Aparecen **Ana (6)** y **Eva (9)**; Luis queda fuera porque tiene un 4.

### 3. Ordenación con `sort_values`

`df.sort_values('stock', ascending=False)` ordena de mayor a menor stock. El orden resultante es B (10), A (3), C (1). El primer registro es **el producto B**, con stock 10. `ascending=False` indica orden descendente.

### 4. Histograma frente a diagrama de barras

`plt.hist(df['nota'], bins=5)` crea un histograma: agrupa valores numéricos en cinco intervalos y muestra cuántas observaciones caen en cada intervalo. Se usa para estudiar la distribución de una variable numérica. Un diagrama de barras se usa normalmente para comparar categorías discretas, como productos, equipos o departamentos.

````python
# Comprobación práctica del bloque C
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

a = np.array([10, 20, 30, 40])
b = a[1:3]
b[0] = 99
print("1) Array original después de modificar la vista:", a)

df_notas = pd.DataFrame({"nombre": ["Ana", "Luis", "Eva"], "nota": [6, 4, 9]})
print("2) Filas con nota >= 5:\n", df_notas[df_notas["nota"] >= 5])

df_stock = pd.DataFrame({"producto": ["A", "B", "C"], "stock": [3, 10, 1]})
print("3) Orden descendente:\n", df_stock.sort_values("stock", ascending=False))

plt.figure(figsize=(6, 4))
plt.hist(df_notas["nota"], bins=5)
plt.title("Distribución de las notas")
plt.xlabel("Nota")
plt.ylabel("Frecuencia")
plt.tight_layout()
plt.show()
````

## Bloque D — Variantes 4, 8, 12, 16 y 20

### 1. Suma con `axis=0`

```python
m = np.array([[1, 2], [3, 4], [5, 6]])
print(m.sum(axis=0))
```

La suma se realiza por columnas (se suman las filas verticalmente): primera columna `1 + 3 + 5 = 9`; segunda columna `2 + 4 + 6 = 12`. La salida es **`[ 9 12]`**. `axis=0` elimina el eje de las filas y conserva un resultado por columna.

### 2. `value_counts`

Para `['rojo', 'azul', 'rojo', 'verde', 'azul', 'rojo']`, el resultado cuenta las apariciones de cada valor y los ordena de mayor a menor frecuencia:

```text
rojo     3
azul     2
verde    1
```

### 3. `dropna` e `inplace=False`

`df2 = df.dropna()` crea un nuevo DataFrame sin las filas que contienen valores nulos. El DataFrame original `df` permanece sin cambios porque `inplace=False` es el comportamiento predeterminado. Con `inplace=True` la modificación se aplica sobre el objeto original y el método devuelve `None`.

### 4. Parámetro `hue` en Seaborn

En `sns.scatterplot(data=df, x='altura', y='peso', hue='sexo')`, `x` y `y` determinan las coordenadas de cada punto; `hue` asigna un color diferente a cada categoría de `sexo` y añade una leyenda. Así se pueden comparar visualmente los grupos en el mismo gráfico.

````python
# Comprobación práctica del bloque D
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

m = np.array([[1, 2], [3, 4], [5, 6]])
print("1) Suma por columnas:", m.sum(axis=0))

serie = pd.Series(["rojo", "azul", "rojo", "verde", "azul", "rojo"])
print("2) Frecuencias:\n", serie.value_counts())

df_con_nulos = pd.DataFrame({"dato": [10, None, 30], "grupo": ["A", "B", "A"]})
df_sin_nulos = df_con_nulos.dropna()  # inplace=False por defecto
print("3) DataFrame nuevo sin nulos:\n", df_sin_nulos)
print("3) DataFrame original, sin modificar:\n", df_con_nulos)

df_personas = pd.DataFrame({
    "altura": [160, 170, 180, 165, 175, 185],
    "peso": [55, 65, 80, 58, 72, 85],
    "sexo": ["F", "M", "M", "F", "F", "M"]
})
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df_personas, x="altura", y="peso", hue="sexo")
plt.title("Relación entre altura y peso por grupo")
plt.xlabel("Altura (cm)")
plt.ylabel("Peso (kg)")
plt.tight_layout()
plt.show()
````
