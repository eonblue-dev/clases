## Bloque 2: Técnicas de Carga de Archivos Locales y Remotos (25 min)

Los datasets en proyectos reales provienen de fuentes muy variadas: servidores web, repositorios públicos, servicios en la nube o archivos locales.

### 2.1 Carga de Archivos CSV (Locales y vía HTTP/HTTPS)

La función `pd.read_csv()` de Pandas es extremadamente flexible. Puede recibir tanto una ruta de archivo local como una dirección URL web.

````python
# Ejemplo 1: Carga directa de un CSV desde una URL pública (Repositorio MIT OCW - Dataset Iris)
url_csv = 'https://ocw.mit.edu/courses/15-097-prediction-machine-learning-and-statistics-spring-2012/89d88c5528513adc4002a1618ce2efb0_iris.csv'

# Leemos el CSV asignando nombres a las columnas
df_iris_url = pd.read_csv(url_csv, names=['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class'])

# Mostramos las primeras filas
display(df_iris_url.head())
````

### 2.2 Uso de Comandos de Shell en Google Colab (`!wget`, `!unzip`, `!gdown`)

En Google Colab, anteponer el signo de exclamación `!` a una celda permite ejecutar **comandos de terminal de Linux/Unix**.

````python
# Ejemplo 2: Descargar un archivo comprimido ZIP directamente desde el repositorio UCI Machine Learning con !wget
!wget -q https://archive.ics.uci.edu/static/public/53/iris.zip -O iris_downloaded.zip

# Descomprimir el archivo ZIP en el entorno local de Colab con !unzip
!unzip -o iris_downloaded.zip

# Leer el fichero 'iris.data' extraído del ZIP
df_iris_uci = pd.read_csv('iris.data', names=['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class'])
display(df_iris_uci.head())
````

### 2.3 Carga de Datos en Formato ARFF (`scipy.io.arff`)

El formato ARFF (*Attribute-Relation File Format*) se utiliza en software clásico como WEKA o repositorios científicos. En Python se lee mediante la librería `scipy`.

````python
from scipy.io import arff

# Intentamos cargar un archivo ARFF (si existe en el directorio de trabajo)
try:
    datos, metadatos = arff.loadarff('iris.arff')
    df_arff = pd.DataFrame(datos)
    
    # En los archivos ARFF las cadenas de texto se importan a veces como objetos 'bytes' (ej: b'Iris-setosa')
    # Convertimos las columnas de tipo objeto/bytes a texto estándar (str/utf-8)
    for col in df_arff.select_dtypes([object]).columns:
        try:
            df_arff[col] = df_arff[col].str.decode('utf-8')
        except Exception:
            pass
            
    display(df_arff.head())
except FileNotFoundError:
    print("Información: Para probar ARFF, asegúrate de subir 'iris.arff' al panel de archivos de la izquierda.")
````

## Bloque 3: Inspección Inicial y Metadatos del DataFrame (15 min)

Una vez cargado un dataset, es imprescindible realizar una exploración de su estructura antes de procesarlo.

````python
# Tomamos como referencia el DataFrame cargado previamente
df = df_iris_url

# 1. Dimensiones (Filas, Columnas)
print(f"Dimensiones del DataFrame (filas, columnas): {df.shape}")
print("-" * 50)

# 2. Tipos de datos por columna
print("Tipos de datos de cada columna:")
print(df.dtypes)
print("-" * 50)

# 3. Resumen estructural completo (tipos, nulos y uso de memoria)
print("Resumen informativo de df.info():")
df.info()
````

### ¿Qué ocurre si el CSV no incluye encabezados (Headers)?

Si leemos un archivo sin cabecera mediante `pd.read_csv()` estándar, Pandas tomará la **primera fila de datos como nombre de las columnas**.

````python
# Demostración: Leer sin especificar 'header' o 'names'
df_sin_header = pd.read_csv(url_csv)
print("ERROR FRECUENTE: Observa cómo la primera fila de datos se convirtió en la cabecera del DataFrame:")
display(df_sin_header.head(3))

print("\nSOLUCIÓN: Usar header=None o pasar la lista de nombres con names=['...']:")
df_con_names = pd.read_csv(url_csv, names=['Sepal_Len', 'Sepal_Wid', 'Petal_Len', 'Petal_Wid', 'Species'])
display(df_con_names.head(3))
````
