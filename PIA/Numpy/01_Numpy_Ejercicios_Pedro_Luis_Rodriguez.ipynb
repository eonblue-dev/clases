## Ejercicio 6

Crea un array de una dimensión de tipo int8. En una de las posiciones pon el valor 1000. ¿Qué ocurre y porqué?

```python
# Creo un array de una dimensión diciéndole que el tipo de dato es int8
array_int8 = np.array([1, 2, 3, 4, 5], dtype=np.int8)
print(array_int8)
print(array_int8.dtype)

# Intento meter el valor 1000 en la posición 2.
# Lo pongo dentro de un try/except porque en las versiones nuevas de NumPy esto da error
# y así el notebook no se para al ejecutarlo entero.
try:
    array_int8[2] = 1000
    print(array_int8)
except OverflowError as error:
    print("Error:", error)
```

**¿Qué ocurre y por qué?**

Un `int8` es un número entero que se guarda en **8 bits** (1 byte). Con 8 bits solo se pueden representar 2⁸ = 256 valores distintos, y como también hay negativos, el rango va de **-128 a 127**. El número 1000 no cabe ahí.

- En las versiones nuevas de NumPy (2.x, la que usamos aquí) NumPy se da cuenta y lanza un `OverflowError` diciendo que 1000 está fuera de los límites de `int8`.
- En versiones antiguas (como la del ejemplo de la teoría) no daba error: el número "daba la vuelta" (*overflow*) y se guardaba un valor raro, **-24**, porque 1000 - 4·256 = -24.

Por eso, como dice la teoría, hay que tener cuidado con el tipo de dato que elegimos: cada tipo tiene su límite numérico.

## Ejercicio 11

Crea una matriz de 3x6 con valores aleatorios. Después muestra por pantalla las siguientes convinaciones:
*   La primera fila entera
*   La primera columna entera
*   Todos los valores entre las columnas 2 y 4
*   Todos los valores de las filas 1 y 2
*   Todos los valores entre las filas 2 y 3 y las columnas 5 y 6

```python
# Matriz de 3 filas y 6 columnas con números aleatorios entre 0 y 1
matriz = np.random.random((3, 6))
print(matriz)
print()

# OJO: en Python las posiciones empiezan en 0.
# Cuando el enunciado dice "fila 1" yo entiendo la primera fila, que en Python es la posición 0.
# En los cortes [inicio:fin] el fin NO se incluye.

# La primera fila entera -> fila 0, todas las columnas
print("Primera fila:")
print(matriz[0, :])
print()

# La primera columna entera -> todas las filas, columna 0
print("Primera columna:")
print(matriz[:, 0])
print()

# Columnas 2, 3 y 4 -> posiciones 1, 2 y 3 -> corte 1:4
print("Columnas de la 2 a la 4:")
print(matriz[:, 1:4])
print()

# Filas 1 y 2 -> posiciones 0 y 1 -> corte 0:2
print("Filas 1 y 2:")
print(matriz[0:2, :])
print()

# Filas 2 y 3 (posiciones 1 y 2) y columnas 5 y 6 (posiciones 4 y 5)
print("Filas 2-3 y columnas 5-6:")
print(matriz[1:3, 4:6])
```

## Ejercicio 13

Crea dos matrices 3x3, una con valores aleatorios y otra de identidad. Realiza las cuatro operaciones básicas entre ellas (suma, resta, multiplicación y división). Por último, calcula la raíz cuadrada de matriz de identidad.

```python
# Matriz 3x3 con valores aleatorios entre 0 y 1
aleatoria = np.random.random((3, 3))
print("Matriz aleatoria:")
print(aleatoria)
print()

# Matriz identidad 3x3 (unos en la diagonal y ceros en el resto)
identidad = np.identity(3)
print("Matriz identidad:")
print(identidad)
print()

# Las operaciones se hacen posición a posición
print("Suma:")
print(aleatoria + identidad)
print()

print("Resta:")
print(aleatoria - identidad)
print()

print("Multiplicación:")
print(aleatoria * identidad)
print()

# OJO: la identidad tiene ceros fuera de la diagonal, así que aquí estamos dividiendo entre 0.
# NumPy no se para: pone "inf" (infinito) en esas posiciones y muestra un aviso (RuntimeWarning).
print("División:")
print(aleatoria / identidad)
print()

# Raíz cuadrada de la identidad: la raíz de 1 es 1 y la de 0 es 0, así que sale la misma matriz
print("Raíz cuadrada de la identidad:")
print(np.sqrt(identidad))
```

## Ejercicio 14

Sin utilizar las funciones std, mean y median de NumPy, programa dichas funcionalidades:

```python
# Para que las funciones sirvan para matrices de cualquier forma,
# primero las paso a una sola dimensión con reshape(tamaño total).

def mi_media(matriz):
    # media = suma de todos los valores / número de valores
    valores = matriz.reshape(matriz.size)
    suma = 0
    for valor in valores:
        suma = suma + valor
    return suma / matriz.size


def mi_desviacion(matriz):
    # desviación típica = raíz cuadrada de la media de (valor - media) al cuadrado
    valores = matriz.reshape(matriz.size)
    media = mi_media(matriz)
    suma_cuadrados = 0
    for valor in valores:
        suma_cuadrados = suma_cuadrados + (valor - media) ** 2
    varianza = suma_cuadrados / matriz.size
    return np.sqrt(varianza)


def mi_mediana(matriz):
    # mediana = el valor del medio cuando los números están ordenados de menor a mayor
    valores = matriz.reshape(matriz.size)
    ordenados = sorted(valores)  # sorted es de Python y devuelve una lista ordenada
    n = len(ordenados)
    mitad = n // 2  # división entera
    if n % 2 == 1:
        # si hay un número impar de valores, la mediana es el del centro
        return ordenados[mitad]
    else:
        # si es par, la mediana es la media de los dos del centro
        return (ordenados[mitad - 1] + ordenados[mitad]) / 2


# Compruebo que dan lo mismo que las funciones de NumPy
prueba = np.random.randint(10, size=(3, 4))
print(prueba)
print()
print("Media      -> mía:", mi_media(prueba), "| NumPy:", np.mean(prueba))
print("Desviación -> mía:", mi_desviacion(prueba), "| NumPy:", np.std(prueba))
print("Mediana    -> mía:", mi_mediana(prueba), "| NumPy:", np.median(prueba))
```

## Ejercicio 15

Compara los tiempos de ejecución entre las funciones que has creado anteriormente  y las de NumPy para matrices de tamaño 1.000.000 x 1.000.000 con valores aleatorios:

**Nota sobre el tamaño:** una matriz de 1.000.000 x 1.000.000 tendría 10¹² elementos (un billón). Cada número `float64` ocupa 8 bytes, así que necesitaría unos **8 TB de memoria RAM**, y el ordenador no puede crearla. Por eso uso una matriz de **1.000 x 1.000**, que tiene 1.000.000 de elementos (el mismo tamaño que el ejemplo de la teoría).

```python
# Uso la librería time como en la teoría para medir cuánto tarda cada función
import time

# 1000 x 1000 = 1.000.000 de valores aleatorios (ver la nota de arriba)
matriz_grande = np.random.random((1000, 1000))

# ---- Media ----
start = time.time()
media_mia = mi_media(matriz_grande)
tiempo_mio = (time.time() - start) * 1000  # en milisegundos

start = time.time()
media_numpy = np.mean(matriz_grande)
tiempo_numpy = (time.time() - start) * 1000

print("MEDIA")
print("  Mía:  ", media_mia, "->", tiempo_mio, "ms")
print("  NumPy:", media_numpy, "->", tiempo_numpy, "ms")
print()

# ---- Desviación típica ----
start = time.time()
desv_mia = mi_desviacion(matriz_grande)
tiempo_mio = (time.time() - start) * 1000

start = time.time()
desv_numpy = np.std(matriz_grande)
tiempo_numpy = (time.time() - start) * 1000

print("DESVIACIÓN TÍPICA")
print("  Mía:  ", desv_mia, "->", tiempo_mio, "ms")
print("  NumPy:", desv_numpy, "->", tiempo_numpy, "ms")
print()

# ---- Mediana ----
start = time.time()
mediana_mia = mi_mediana(matriz_grande)
tiempo_mio = (time.time() - start) * 1000

start = time.time()
mediana_numpy = np.median(matriz_grande)
tiempo_numpy = (time.time() - start) * 1000

print("MEDIANA")
print("  Mía:  ", mediana_mia, "->", tiempo_mio, "ms")
print("  NumPy:", mediana_numpy, "->", tiempo_numpy, "ms")
```

**Conclusión:** los resultados son prácticamente iguales (puede variar algún decimal del final por cómo se redondean las sumas), pero las funciones de NumPy son muchísimo más rápidas. Esto pasa porque mis funciones recorren los valores uno a uno con un bucle `for` de Python, mientras que NumPy hace los cálculos en código C optimizado, como se explica en la introducción y en el apartado de optimización de la teoría.
