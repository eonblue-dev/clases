## 2. Qué es Seaborn y cómo se organiza

Seaborn es una biblioteca de visualización estadística de Python construida sobre Matplotlib. Permite crear gráficos estadísticos con poco código y trabajar directamente con columnas de DataFrames de pandas.

### Diferencia general entre Matplotlib y Seaborn

| Característica | Matplotlib | Seaborn |
|---|---|---|
| Nivel de abstracción | Bajo: control detallado | Alto: funciones estadísticas preparadas |
| Estilo predeterminado | Más básico | Temas visuales integrados |
| Integración con pandas | Compatible | Diseñada para DataFrames |
| Estadística | Se construye manualmente | Incluye estimaciones, KDE e intervalos |
| Paneles múltiples | Se organizan manualmente | Tiene funciones y grids específicos |
| Uso habitual | Personalización avanzada | Exploración y visualización estadística |

En un proyecto real es habitual **combinar ambas bibliotecas**: Seaborn para crear el gráfico y Matplotlib para ajustar sus detalles.

### Tres grupos de funciones

1. **Axes-level:** dibujan sobre un único eje de Matplotlib. Ejemplos: `scatterplot()`, `histplot()` y `boxplot()`.
2. **Figure-level:** crean una figura completa y pueden dividir los datos en paneles. Ejemplos: `relplot()`, `displot()` y `catplot()`.
3. **Funciones especializadas:** generan composiciones específicas. Ejemplos: `heatmap()`, `pairplot()` y `jointplot()`.

### Tipos de análisis visual

- **Univariante:** estudia una variable.
- **Bivariante:** estudia la relación entre dos variables.
- **Multivariante:** estudia tres o más variables o varias relaciones a la vez.

## 4. Funciones Axes-level y Figure-level

Esta distinción es importante porque determina cómo se organiza y personaliza el gráfico.

### 4.1. Dos gráficos Axes-level en una misma figura

`plt.subplots()` crea una figura y varios ejes. Después, se entrega cada eje mediante el parámetro `ax`.

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

tips = sns.load_dataset("tips")

# Cada función dibuja sobre un eje concreto
sns.histplot(data=tips, x="total_bill", ax=axes[0])
sns.boxplot(data=tips, x="day", y="total_bill", ax=axes[1])

plt.tight_layout()
plt.show()
````

En este caso, `histplot()` y `boxplot()` no crean una figura independiente: dibujan sobre `axes[0]` y `axes[1]`.

### 4.2. Función Figure-level con `displot()`

Las funciones Figure-level pueden dividir automáticamente los datos en paneles mediante `col` o `row`.

````python
tips = sns.load_dataset("tips")

# Un panel para cada valor de la columna "time"
g = sns.displot(data=tips, x="total_bill", col="time", kde=True)

# En funciones Figure-level se pueden configurar las etiquetas desde el objeto FacetGrid
g.set_axis_labels("Importe de la factura", "Frecuencia")
g.set_titles("Comida: {col_name}")

plt.show()
````

**Diferencia importante:** las funciones Figure-level suelen devolver un objeto como `FacetGrid`, no un eje individual. Para cambiar etiquetas y títulos es conveniente utilizar los métodos del objeto devuelto, como `set_axis_labels()` y `set_titles()`.

## 5. Gráficos relacionales

Los gráficos relacionales muestran cómo se relacionan dos o más variables. Los más habituales son el **scatter plot** y el **line plot**.

### 5.1. Scatter plot — gráfico de dispersión

Se utiliza para estudiar la relación entre dos variables numéricas. Cada punto representa una observación.

#### Ejemplo básico

````python
tips = sns.load_dataset("tips")

sns.scatterplot(x="total_bill", y="tip", data=tips)
plt.title("Relación entre factura y propina")
plt.xlabel("Importe de la factura")
plt.ylabel("Propina")
plt.show()
````

**Interpretación:** permite observar si, a medida que aumenta una variable, la otra tiende también a aumentar o disminuir. También ayuda a detectar grupos y valores atípicos.

#### Personalizar con `hue` y `size`

- `hue`: asigna colores según una variable categórica.
- `size`: modifica el tamaño de los puntos según una variable.
- `sizes`: establece el rango de tamaños de los puntos.

````python
tips = sns.load_dataset("tips")

# Personalizar el gráfico de dispersión
sns.scatterplot(
    x="total_bill",
    y="tip",
    hue="sex",
    size="size",
    sizes=(50, 200),
    data=tips
)

# Añadir etiquetas y título
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Relationship between Total Bill and Tip")
plt.show()
````

**Qué debes recordar:** `hue` y `size` permiten incorporar información adicional sin añadir nuevos ejes. Conviene no utilizar demasiadas categorías o tamaños porque el gráfico puede resultar difícil de interpretar.

### 5.2. Line plot — gráfico de líneas

Se utiliza para representar tendencias, especialmente cuando una variable tiene un orden natural, como el tiempo.

#### Ejemplo básico con `fmri`

````python
fmri = sns.load_dataset("fmri")

sns.lineplot(x="timepoint", y="signal", data=fmri)
plt.title("Señal en función del tiempo")
plt.xlabel("Punto temporal")
plt.ylabel("Intensidad de la señal")
plt.show()
````

#### Personalizar con `hue`, `style`, `markers` y `dashes`

- `hue`: diferencia grupos mediante colores.
- `style`: diferencia grupos mediante el estilo de línea.
- `markers=True`: muestra marcadores en los puntos.
- `dashes=False`: evita los trazos discontinuos.

````python
fmri = sns.load_dataset("fmri")

# Personalizar el gráfico de líneas
sns.lineplot(
    x="timepoint",
    y="signal",
    hue="event",
    style="region",
    markers=True,
    dashes=False,
    data=fmri
)

plt.xlabel("Timepoint")
plt.ylabel("Signal Intensity")
plt.title("Changes in Signal Intensity over Time")
plt.show()
````

**Consejo de análisis:** una línea puede sugerir continuidad. Utilízala cuando el orden de los valores tenga sentido (por ejemplo, tiempo, distancia o etapas). Para categorías sin orden natural, suele ser mejor un gráfico de puntos o barras.

## 6. Gráficos categóricos

Los gráficos categóricos permiten comparar una variable numérica entre grupos, clases o categorías.

### 6.1. Bar plot — gráfico de barras

`barplot()` representa un estadístico de una variable numérica para cada categoría. Por defecto, normalmente muestra la **media** y su intervalo de incertidumbre.

#### Ejemplo básico con `titanic`

````python
titanic = sns.load_dataset("titanic")

sns.barplot(x="class", y="fare", data=titanic)
plt.title("Tarifa media por clase")
plt.xlabel("Clase")
plt.ylabel("Tarifa")
plt.show()
````

#### Personalizar con `hue`, `errorbar` y `palette`

- `hue`: separa cada categoría en grupos adicionales.
- `errorbar=None`: elimina las barras de incertidumbre.
- `palette`: selecciona una paleta de colores.

````python
titanic = sns.load_dataset("titanic")

sns.barplot(
    x="class",
    y="fare",
    hue="sex",
    errorbar=None,
    palette="muted",
    data=titanic
)

plt.xlabel("Class")
plt.ylabel("Fare")
plt.title("Average Fare by Class and Gender on the Titanic")
plt.show()
````

**Importante:** una barra representa un resumen estadístico, no necesariamente el valor de cada observación. Si necesitas representar todos los registros, considera un `stripplot()` o `swarmplot()`.

### 6.2. Box plot — diagrama de caja y bigotes

Un box plot resume la distribución de una variable numérica:

- Línea central: mediana.
- Caja: desde el primer cuartil (Q1) hasta el tercer cuartil (Q3).
- Altura de la caja: rango intercuartílico (IQR = Q3 − Q1).
- Bigotes: se extienden hasta los valores considerados no atípicos según la regla del gráfico.
- Puntos fuera de los bigotes: posibles valores atípicos.

#### Ejemplo básico

````python
tips = sns.load_dataset("tips")

sns.boxplot(x="day", y="total_bill", data=tips)
plt.title("Distribución de las facturas por día")
plt.xlabel("Día")
plt.ylabel("Importe de la factura")
plt.show()
````

#### Personalizar con `hue`, `palette`, `linewidth` y `fliersize`

````python
# Cargar el dataset tips
tips = sns.load_dataset("tips")

# Crear un box plot por día y momento de comida
sns.boxplot(
    x="day",
    y="total_bill",
    hue="time",
    data=tips,
    palette="Set3",
    linewidth=1.5,
    fliersize=4
)

plt.title("Box Plot of Total Bill by Day and Meal Time")
plt.xlabel("Day of the Week")
plt.ylabel("Total Bill ($)")
plt.show()
````

### 6.3. Violin plot — gráfico de violín

Combina información de un box plot con una estimación de densidad KDE. La anchura indica dónde se concentran más observaciones: las zonas anchas representan mayor densidad.

#### Ejemplo con `iris`

````python
iris = sns.load_dataset("iris")

sns.violinplot(x="species", y="petal_length", data=iris)
plt.title("Longitud del pétalo según la especie")
plt.xlabel("Especie")
plt.ylabel("Longitud del pétalo")
plt.show()
````

**Cuándo utilizarlo:**
- Box plot: cuando quieres comparar medianas, cuartiles y posibles valores atípicos.
- Violin plot: cuando además quieres observar la forma de la distribución.
- Ambos: cuando quieres comparar la distribución numérica entre categorías.

## 7. Gráficos de distribución

Los gráficos de distribución muestran cómo se reparten los valores de una variable numérica.

### 7.1. Histograma — `histplot()`

Divide los valores en intervalos (*bins*) y muestra cuántas observaciones caen en cada intervalo.

#### Ejemplo básico

````python
iris = sns.load_dataset("iris")

sns.histplot(x="petal_length", data=iris)
plt.title("Distribución de la longitud de los pétalos")
plt.xlabel("Longitud del pétalo")
plt.ylabel("Frecuencia")
plt.show()
````

#### Personalizar el histograma

- `bins=20`: utiliza 20 intervalos.
- `kde=True`: añade una curva de densidad estimada.
- `color="green"`: cambia el color del histograma.

````python
iris = sns.load_dataset("iris")

sns.histplot(
    data=iris,
    x="petal_length",
    bins=20,
    kde=True,
    color="green"
)

plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")
plt.title("Distribution of Petal Lengths in Iris Flowers")
plt.show()
````

**Cómo interpretar un histograma:**
- La forma ayuda a identificar asimetría, concentración y posibles varios grupos.
- Un número muy bajo de intervalos puede ocultar detalles.
- Un número muy alto puede producir un gráfico irregular.
- Compara siempre distribuciones con una escala y unos intervalos adecuados.

### 7.2. KDE — Kernel Density Estimation

Un KDE estima la densidad de una variable continua y la representa mediante una curva suave. Es una alternativa o complemento al histograma.

#### Ejemplo básico

````python
tips = sns.load_dataset("tips")

sns.kdeplot(data=tips, x="total_bill")
plt.title("Densidad del importe de las facturas")
plt.xlabel("Importe de la factura")
plt.ylabel("Densidad")
plt.show()
````

#### KDE personalizado por momento de comida

- `hue="time"`: crea curvas diferenciadas por comida.
- `fill=True`: rellena el área bajo la curva.
- `alpha=0.6`: controla la transparencia.
- `linewidth=1.5`: establece el grosor de la línea.

````python
# Cargar el dataset tips
tips = sns.load_dataset("tips")

# Crear una curva de densidad del importe de las facturas
# hue diferencia las comidas y fill rellena el área bajo cada curva
sns.kdeplot(
    data=tips,
    x="total_bill",
    hue="time",
    fill=True,
    alpha=0.6,
    linewidth=1.5
)

plt.title("Density Plot of Total Bill by Meal Time")
plt.xlabel("Total Bill ($)")
plt.ylabel("Density")
plt.show()
````

**Precaución:** el KDE es una estimación suavizada, no una representación exacta de cada observación. La forma puede cambiar según el ancho de banda utilizado.

## 8. Gráficos matriciales y multivariantes

Estos gráficos ayudan a visualizar relaciones entre varias variables o a comparar muchos pares de variables.

### 8.1. Heatmap — mapa de calor

Un heatmap representa una matriz mediante colores. Es frecuente utilizarlo para visualizar correlaciones.

#### Ejemplo: matriz de correlación de `tips`

````python
tips = sns.load_dataset("tips")

# Seleccionar únicamente las columnas numéricas
# y calcular su matriz de correlación
corr = tips.select_dtypes(include="number").corr()

# Mostrar las correlaciones con anotaciones
sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Matriz de correlación de Tips")
plt.show()
````

**Interpretación de la correlación:**
- Cerca de `+1`: relación lineal positiva.
- Cerca de `-1`: relación lineal negativa.
- Cerca de `0`: relación lineal débil o inexistente.

La correlación no demuestra causalidad.

#### Ejemplo: heatmap con el dataset `flights`

Para preparar los datos para el mapa de calor, se utiliza `pivot()`:

- `index="month"`: filas por mes.
- `columns="year"`: columnas por año.
- `values="passengers"`: valores que rellenan la matriz.

````python
# Cargar el dataset flights
flights = sns.load_dataset("flights")

# Transformar la tabla para crear una matriz mes x año
flights_pivot = flights.pivot(
    index="month",
    columns="year",
    values="passengers"
)

# Crear el mapa de calor
sns.heatmap(flights_pivot, cmap="Blues", annot=True, fmt="d")

plt.title("Passengers per month")
plt.xlabel("Year")
plt.ylabel("Month")
plt.show()
````

### 8.2. Pair plot — matriz de relaciones

`pairplot()` crea una matriz de gráficos para estudiar las relaciones entre varias columnas numéricas.

- Fuera de la diagonal: gráficos de dispersión entre pares de variables.
- En la diagonal: distribución individual de cada variable.

#### Ejemplo básico

````python
# Cargar el dataset iris
iris = sns.load_dataset("iris")

# Crear la matriz de gráficos
sns.pairplot(data=iris)

plt.show()
````

#### Personalizar con `hue`, `diag_kind` y `palette`

- `hue="species"`: colorea los puntos según la especie.
- `diag_kind="kde"`: representa la diagonal mediante curvas KDE.
- `palette="husl"`: selecciona la paleta de colores.

El artículo utiliza `plt.title()` después de `pairplot()`. Para colocar un título general de forma fiable, es preferible utilizar el objeto devuelto por `pairplot()`, como en la segunda celda.

````python
# Cargar el dataset iris
iris = sns.load_dataset("iris")

# Crear pair plot personalizado
g = sns.pairplot(
    data=iris,
    hue="species",
    diag_kind="kde",
    palette="husl"
)

# Título general de la figura (forma recomendada)
g.fig.suptitle("Iris Dataset Pair Plot", y=1.02)

plt.show()
````

### 8.3. Joint plot — gráfico conjunto

Un `jointplot()` combina:
- Un gráfico principal en el centro (por defecto, dispersión).
- Un histograma marginal en la parte superior.
- Otro histograma marginal en el lateral.

Permite estudiar simultáneamente la relación entre dos variables y sus distribuciones individuales.

````python
# Cargar el dataset iris
iris = sns.load_dataset("iris")

# Gráfico conjunto de longitud y anchura del sépalo
sns.jointplot(
    x="sepal_length",
    y="sepal_width",
    data=iris
)

plt.show()
````

### 8.4. FacetGrid — paneles por categorías

`FacetGrid` crea un panel por cada valor distinto de una variable categórica. Así se puede comparar la misma visualización entre grupos.

En el ejemplo, se crea un histograma de `total_bill` para cada día.

````python
# Cargar el dataset tips
tips = sns.load_dataset("tips")

# Crear un FacetGrid con un panel para cada día
g = sns.FacetGrid(tips, col="day")

# Dibujar un histograma de total_bill en cada panel
g.map(sns.histplot, "total_bill")

g.set_axis_labels("Importe de la factura", "Frecuencia")
g.set_titles("Día: {col_name}")

plt.show()
````

**Diferencia entre `FacetGrid` y `pairplot`:**
- `FacetGrid`: tú decides qué gráfico dibujar y cómo dividir los datos en paneles.
- `pairplot`: crea automáticamente una matriz de relaciones entre variables numéricas.

## 9. Temas y estilos

### Estilos disponibles

- `darkgrid`
- `whitegrid`
- `dark`
- `white`

### Cambiar el contexto con `context`

El contexto ajusta el tamaño de fuentes, líneas y otros elementos:

- `paper`: elementos pequeños, adecuado para figuras compactas.
- `notebook`: escala predeterminada para notebooks.

````python
sns.set_theme(style="whitegrid", context="talk")

tips = sns.load_dataset("tips")
sns.scatterplot(data=tips, x="total_bill", y="tip")
plt.title("Contexto talk")
plt.show()

# Restaurar el contexto habitual para las siguientes celdas
sns.set_theme(style="whitegrid", context="notebook")
````

## 10. Personalización de gráficos

Seaborn crea la visualización estadística y Matplotlib permite ajustar su presentación.

### 10.1. Cambiar la paleta de colores

`palette` define los colores utilizados por grupos representados mediante `hue`.

````python
# Cargar el dataset de ejemplo
tips = sns.load_dataset("tips")

# Crear un scatter plot con una paleta de colores
sns.scatterplot(
    x="total_bill",
    y="tip",
    hue="day",
    data=tips,
    palette="Set2"
)

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()
````

**Paletas que puedes probar:** `deep`, `muted`, `bright`, `pastel`, `dark`, `colorblind`, `Set1`, `Set2`, `Set3`, `viridis`, `magma` y `coolwarm`.

### 10.2. Ajustar el tamaño de la figura

`plt.figure(figsize=(ancho, alto))` establece el tamaño de la figura en pulgadas.

````python
# Cargar el dataset iris
iris = sns.load_dataset("iris")

# Crear una figura de 8 x 6 pulgadas
plt.figure(figsize=(8, 6))

sns.violinplot(
    x="species",
    y="petal_length",
    data=iris
)

plt.title("Petal Length Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")
plt.show()
````

### 10.3. Añadir anotaciones

`plt.text(x, y, texto)` escribe texto en una posición concreta del sistema de coordenadas del gráfico. Es útil para señalar zonas o valores importantes.

````python
# Cargar el dataset diamonds
diamonds = sns.load_dataset("diamonds")

# Crear un gráfico de dispersión
sns.scatterplot(x="carat", y="price", data=diamonds)

# Añadir anotaciones en coordenadas del gráfico
plt.text(
    1,
    18000,
    "Large, Expensive Diamonds",
    fontsize=12,
    color="red"
)

plt.text(
    2.5,
    5000,
    "Small, Affordable Diamonds",
    fontsize=12,
    color="blue"
)

plt.title("Diamond Prices by Carat")
plt.xlabel("Carat (ct)")
plt.ylabel("Price ($)")
plt.show()
````

### Soluciones orientativas

````python
# SOLUCIÓN 1 — Scatter plot
tips = sns.load_dataset("tips")
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="day")
plt.title("Factura y propina según el día")
plt.show()

# SOLUCIÓN 2 — Histograma + KDE
iris = sns.load_dataset("iris")
sns.histplot(data=iris, x="sepal_length", bins=25, kde=True, hue=None)
plt.title("Distribución de la longitud del sépalo")
plt.show()

# SOLUCIÓN 2, comparación por especie
sns.histplot(data=iris, x="sepal_length", bins=25, kde=True, hue="species")
plt.show()

# SOLUCIÓN 3 — Box plot
tips = sns.load_dataset("tips")
sns.boxplot(data=tips, x="day", y="total_bill", hue="sex")
plt.title("Importe de la factura por día y sexo")
plt.show()

# SOLUCIÓN 4 — Correlación sin comparar una variable consigo misma
corr = tips.select_dtypes(include="number").corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.show()

corr_sin_diagonal = corr.where(~np.eye(corr.shape[0], dtype=bool))
pares = corr_sin_diagonal.abs().unstack().dropna().sort_values(ascending=False)
pares = pares[~pares.index.map(lambda par: par[0] == par[1])]
print("Pareja con mayor correlación absoluta:")
print(pares.head(2))

# SOLUCIÓN 5 — FacetGrid
g = sns.FacetGrid(tips, col="time")
g.map(sns.histplot, "total_bill")
g.set_axis_labels("Importe de la factura", "Frecuencia")
plt.show()
````
