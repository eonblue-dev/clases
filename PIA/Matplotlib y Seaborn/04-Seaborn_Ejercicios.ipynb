# Ejercicios Seaborn

Relación de ejercicios para practicar distintos tipos de gráficas con Seaborn. Se usa el dataset **iris** (el mismo de la teoría) y el de **COVID-19** de la carpeta para las gráficas en el tiempo. Completa los ejercicios en orden:

---
## Ejercicio 1

Importa las librerías necesarias para trabajar con Seaborn (recuerda que Seaborn necesita Matplotlib, y que vamos a leer los datos con Pandas). Activa también el estilo por defecto de Seaborn:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# para que salga el fondo de seaborn y no el blanco de matplotlib
sns.set()
```

---
## Ejercicio 2

Carga el dataset `iris.csv` y muestra sus primeras filas, su tamaño y si tiene valores nulos:

```python
# el csv de iris lo tengo en la carpeta "Ejercicios variados"
iris = pd.read_csv("../Ejercicios variados/iris.csv")

print(iris.head(10))
print(iris.shape)
print(iris.isnull().sum())
```

---
## Ejercicio 3

Muestra un **diagrama de enjambre** (`swarmplot`) con la anchura del sépalo (`sepal.width`) de cada variedad:

```python
sns.swarmplot(x = "variety", y = "sepal.width", data = iris, size = 3)

plt.title("Anchura del sépalo por variedad")
plt.show()
```

---
## Ejercicio 4

Muestra un **gráfico de barras** (`barplot`) con la longitud media del pétalo de cada variedad:

```python
# barplot ya calcula la media él solo, y la rayita negra es el intervalo de confianza
sns.barplot(x = "variety", y = "petal.length", data = iris)

plt.title("Longitud media del pétalo por variedad")
plt.xlabel("Variedad")
plt.ylabel("Longitud del pétalo")
plt.show()
```

---
## Ejercicio 5

Muestra un **gráfico de conteo** (`countplot`) con cuántas flores hay de cada variedad:

```python
sns.countplot(x = "variety", data = iris)

plt.title("Número de flores de cada variedad")
plt.show()

# sale 50 de cada una, el dataset está equilibrado
```

---
## Ejercicio 6

Muestra un **histograma** (`histplot`) de la longitud del pétalo, coloreando cada variedad de un color distinto (parámetro `hue`):

```python
sns.histplot(x = "petal.length", data = iris, hue = "variety", bins = 20)

plt.title("Histograma de la longitud del pétalo")
plt.show()
```

---
## Ejercicio 7

Repite el histograma anterior pero añadiendo la **curva de densidad** (`kde`):

```python
sns.histplot(x = "petal.length", data = iris, hue = "variety", bins = 20, kde = True)

plt.title("Histograma con curva de densidad")
plt.show()
```

---
## Ejercicio 8

Muestra un **gráfico de dispersión** (`scatterplot`) de la longitud del pétalo frente a su anchura, con un color por variedad:

```python
sns.scatterplot(x = "petal.length", y = "petal.width", data = iris, hue = "variety")

plt.title("Pétalo: longitud vs anchura")
plt.show()

# se ve que la setosa tiene los pétalos mucho más pequeños que las otras dos
```

---
## Ejercicio 9

Muestra un **diagrama de cajas** (`boxplot`) de la longitud del sépalo para cada variedad:

```python
sns.boxplot(x = "variety", y = "sepal.length", data = iris)

plt.title("Longitud del sépalo por variedad")
plt.show()
```

---
## Ejercicio 10

Muestra un **diagrama de violín** (`violinplot`) con los mismos datos que el ejercicio anterior. ¿Qué información extra te da respecto al boxplot?

```python
sns.violinplot(x = "variety", y = "sepal.length", data = iris)

plt.title("Longitud del sépalo por variedad (violín)")
plt.show()

# el violín además de la caja enseña la forma de la distribución (donde hay más flores)
```

---
## Ejercicio 11

Muestra un **mapa de calor** (`heatmap`) con la correlación entre las columnas numéricas del dataset. Muestra el valor dentro de cada casilla:

```python
# la correlación solo se puede calcular con columnas numéricas, así que quito "variety"
numericas = iris.drop(columns = "variety")
correlacion = numericas.corr()
print(correlacion)

sns.heatmap(correlacion, annot = True, cmap = "coolwarm")

plt.title("Correlación entre variables")
plt.show()
```

---
## Ejercicio 12

Muestra una **matriz de gráficos** (`pairplot`) con todas las combinaciones de variables, coloreando por variedad:

```python
# esta tarda un poco más porque hace muchas gráficas a la vez
sns.pairplot(iris, hue = "variety")
plt.show()
```

## Ejercicio 13

Muestra un **gráfico conjunto** (`jointplot`) de la longitud y la anchura del sépalo:

```python
sns.jointplot(x = "sepal.length", y = "sepal.width", data = iris, hue = "variety")
plt.show()
```

---
## Ejercicio 14

Muestra un **gráfico de dispersión con recta de regresión** (`regplot`) de la longitud del pétalo frente a la longitud del sépalo:

```python
sns.regplot(x = "sepal.length", y = "petal.length", data = iris, color = "green")

plt.title("Longitud del sépalo vs longitud del pétalo")
plt.show()

# la recta sube, o sea que cuanto más largo el sépalo más largo el pétalo
```

---
## Ejercicio 15

Carga el dataset `03-nacional_covid19.csv` y muestra un **diagrama de líneas** (`lineplot`) con la evolución de los fallecimientos:

```python
covid = pd.read_csv("03-nacional_covid19.csv")

# paso la fecha a tipo fecha para que el eje X salga bien ordenado
covid["fecha"] = pd.to_datetime(covid["fecha"])

sns.lineplot(x = "fecha", y = "fallecimientos", data = covid)

plt.title("Fallecimientos acumulados por COVID-19 en España")
plt.xticks(rotation = 45)
plt.show()
```

---
## Ejercicio 16

Con el mismo dataset, muestra **varias líneas en la misma gráfica** (hospitalizados, ingresos en UCI y fallecimientos):

```python
sns.lineplot(x = "fecha", y = "hospitalizados", data = covid, label = "Hospitalizados")
sns.lineplot(x = "fecha", y = "ingresos_uci", data = covid, label = "UCI")
sns.lineplot(x = "fecha", y = "fallecimientos", data = covid, label = "Fallecimientos")

plt.title("Evolución del COVID-19 en España (acumulado)")
plt.ylabel("Personas")
plt.xticks(rotation = 45)
plt.legend()
plt.show()

# hay huecos en las líneas porque faltan datos (nulos) en algunas fechas
```

## Ejercicio 17

Para terminar, junta **cuatro gráficas distintas** en una misma figura usando `plt.subplots` y el parámetro `ax` de Seaborn:

```python
fig, ejes = plt.subplots(2, 2, figsize = (12, 8))

sns.countplot(x = "variety", data = iris, ax = ejes[0, 0])
ejes[0, 0].set_title("Conteo")

sns.boxplot(x = "variety", y = "petal.width", data = iris, ax = ejes[0, 1])
ejes[0, 1].set_title("Cajas")

sns.histplot(x = "sepal.width", data = iris, ax = ejes[1, 0])
ejes[1, 0].set_title("Histograma")

sns.scatterplot(x = "sepal.length", y = "petal.width", data = iris, hue = "variety", ax = ejes[1, 1])
ejes[1, 1].set_title("Dispersión")

# para que no se pisen los títulos
plt.tight_layout()
plt.show()
```
