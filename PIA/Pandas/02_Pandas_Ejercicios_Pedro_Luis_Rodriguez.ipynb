## Ejercicio 3

Crea una matriz de NumPy con al menos 3 filas y 3 columnas. A partir de esta, crea un dataframe, teniendo en cuenta que la primera fila serán los nombres de las columnas y la primera columna los índices de las filas. Muestra el DataFrame por pantalla:

```python
# Creo una matriz de NumPy de 3x3.
# La primera fila tiene los nombres de las columnas y la primera columna los nombres de las filas.
# (NumPy guarda todo como texto porque en una matriz todos los datos son del mismo tipo)
data = np.array([['','Nombre','Edad'],['F1','Mikel',45],['F2','Pedro',25]])
print('Numpy data: ')
print(data)

# Creo el DataFrame como en la teoría:
#   data[1:,1:] -> los datos: desde la fila 1 y la columna 1 hasta el final
#   data[1:,0]  -> el índice de las filas: la columna 0 desde la fila 1
#   data[0,1:]  -> los nombres de las columnas: la fila 0 desde la columna 1
pdata = pd.DataFrame(data=data[1:,1:],index=data[1:,0],columns=data[0,1:])
print('Pandas data: ')
print(pdata)
```

Del DataFrame creado en el ejercicio anterior, muestra su forma, índice, número de filas, media por columnas, valores máximos y mínimos por columnas, mediana por columnas y la desviación típica por columnas:

## Ejercicio 6

Crea un DataFrame a partir de una matriz con valores definidos por tí. Estos valores deberán incluir valores nulos. Luego muestra la cantidad de valores nulos por pantalla y modifica los valores nulos por la media que tenga dicha columna sin incluir los nulos. Muestra el resultado por pantalla:

```python
# 1. Crear un DataFrame con valores definidos por nosotros, incluyendo nulos (np.nan)
datos = [
    [10, 5, np.nan],
    [20, np.nan, 15],
    [np.nan, 8, 25],
    [40, 2, 35],
    [50, 6, np.nan]
]

df = pd.DataFrame(datos, columns=['Columna_A', 'Columna_B', 'Columna_C'])

print("--- DataFrame Original con Valores Nulos ---")
print(df)
print("\n" + "="*45 + "\n")

# 2. Mostrar la cantidad de valores nulos por columna
nulos = df.isnull().sum()
print("--- Cantidad de Valores Nulos por Columna ---")
print(nulos)
print("\n" + "="*45 + "\n")

# 3. Reemplazar los valores nulos por la media de cada columna
# (el método .mean() de Pandas omite los nulos por defecto automáticamente: skipna=True)
df_limpio = df.fillna(df.mean())

# 4. Mostrar el resultado final
print("--- DataFrame con los Nulos Reemplazados por la Media ---")
print(df_limpio)
```

## Ejercicio 9

Investiga qué es la correlación entre columnas. Cuando lo hayas entendido, crea una matriz 10x3 con datos manuales que entre la primera y segunda columna de una correlación de 1 en ambos sentidos. Muestra la matriz y su correlación entre columnas por pantalla:

**¿Qué es la correlación?**

La correlación es un número entre **-1 y 1** que dice si dos columnas cambian a la vez siguiendo una línea recta:

- **1**: cuando una sube, la otra también sube, siempre en la misma proporción.
- **-1**: cuando una sube, la otra baja, siempre en la misma proporción.
- **0**: no hay relación en línea recta entre ellas.

Además, la correlación de A con B es la misma que la de B con A. Por eso la matriz de `corr()` es simétrica: si A y B tienen correlación 1, sale un 1 tanto en la casilla (A, B) como en la (B, A). Eso es lo que entiendo por "en ambos sentidos".

Para conseguir una correlación de 1 entre la primera y la segunda columna, hago que la segunda sea siempre **el doble** de la primera. La tercera columna tiene valores puestos a mano sin ningún patrón.

```python
# Matriz 10x3 hecha a mano:
#   columna 0 -> números del 1 al 10
#   columna 1 -> el doble de la columna 0 (por eso la correlación será 1)
#   columna 2 -> números puestos sin ningún orden
matriz_corr = np.array([[1, 2, 7],
                        [2, 4, 3],
                        [3, 6, 9],
                        [4, 8, 1],
                        [5, 10, 5],
                        [6, 12, 8],
                        [7, 14, 2],
                        [8, 16, 6],
                        [9, 18, 4],
                        [10, 20, 10]])

# Lo paso a DataFrame y le pongo nombre a las columnas para que se lea mejor
df_corr = pd.DataFrame(matriz_corr, columns=['A', 'B', 'C'])
print('Matriz:')
print(df_corr)
print()

# corr() calcula la correlación entre todas las columnas (como en la teoría)
print('Correlación entre columnas:')
print(df_corr.corr())
```

**Pasos previos (fuera del notebook):**

1. Instalar MySQL Server (en este ordenador todavía no está instalado).
2. Crear el esquema: `CREATE SCHEMA BDIA;`
3. Ejecutar dentro de `BDIA` las sentencias del fichero `02-Pandas_Tabla_MySQL.sql` (el fichero no está en la carpeta, hay que descargarlo).
4. Instalar el conector de Python para MySQL: `pip install pymysql`. `sqlalchemy` ya está instalado.

La teoría solo dice que Pandas puede importar desde una base de datos con funciones parecidas a `read_csv`. Para bases de datos esa función es `read_sql`. Hay que cambiar el usuario, la contraseña y el nombre de la tabla por los vuestros.

```python
# create_engine crea la "conexión" que Pandas necesita para hablar con la base de datos
from sqlalchemy import create_engine

# Datos de conexión
usuario = 'root'
contraseña = 'root'
servidor = 'localhost'   # la base de datos está en mi propio ordenador
puerto = 3306            # puerto por defecto de MySQL
esquema = 'BDIA'         # el esquema que hemos creado

# La dirección tiene la forma: mysql+pymysql://usuario:contraseña@servidor:puerto/esquema
conexion = create_engine(f'mysql+pymysql://{usuario}:{contraseña}@{servidor}:{puerto}/{esquema}')

# read_sql ejecuta una consulta SQL y devuelve el resultado como DataFrame.
# CAMBIAR "nombre_tabla" por el nombre de la tabla que crea el fichero .sql
df_mysql = pd.read_sql('SELECT * FROM nombre_tabla', conexion)

# Muestro los datos
print(df_mysql)
```
