## Primeros pasos

Para crear un gráfico con matplotlib es habitual seguir los siguientes pasos:


1.   **Importar** el módulo **pyplot**.
2.   **Definir la figura que contendrá el gráfico**, que es la region (ventana o página) donde se dibujará y los ejes sobre los que se dibujarán los datos. Para ello se utiliza la `función subplots()`.
3.   **Dibujar los datos sobre los ejes**. Para ello se utilizan distintas funciones dependiendo del tipo de gráfico que se quiera.
4.   **Personalizar el gráfico**. Para ello existen multitud de funciones que permiten añadir un título, una leyenda, una rejilla, cambiar colores o personalizar los ejes.
5.   **Guardar el gráfico**. Para ello se utiliza la `función savefig()`.
6.   **Mostrar el gráfico**. Para ello se utiliza la `función show()`.


Una de las características de **Matplotlib** es la facilidad con la que se puede comenzar a trazar gráficas, vea el siguiente código:

La instrucción anterior produce la gráfica mostrada en la figura, a la función **`plot()`** únicamente se le pasó como argumento una lista de valores numéricos.

Otro ejemplo ilustrativo es el siguiente, donde vemos cómo mostrar rápidamente por pantalla una gráfica con datos aleatorios inventados por nosotros. Después entraremos a analizar y probar distintas combinaciones de parámetros:

````python
a = [1, 2, 3, 4, 5]
b = [5, 1, 3, 2, 4]

# Primero añadiremos los datos al gráfico
plt.plot(a,b,label='línea')

# Si queremos que aparezca la leyenda "línea" tendremos que añadir esta intrucción
plt.legend() 

# Después lo mostraremos por pantalla
plt.show()
````

A la misma gráfica podremos indicarle el color de la línea así como su grosor entre otros:

````python
a = [1, 2, 3, 4, 5]
b = [5, 1, 3, 2, 4]

# Primero añadiremos los datos al gráfico
plt.plot(a,b,label='particula 1', color='red', linewidth=1 , linestyle='--', marker='o', markersize=6, markerfacecolor='red')

# Si queremos que aparezca la leyenda "línea" tendremos que añadir esta intrucción
plt.legend() 

#Marcador (3,3)
plt.annotate(
    'P3 (3, 3)',
    xy=(3, 3),
    xytext=(3, 3.3),
    color='blue'
)

plt.xticks(range(0, 6, 2))  # Cada 2 unidades en X
plt.yticks(range(0, 6, 2))  # Cada 2 unidades en Y

plt.grid(ls='--', alpha=0.3, linewidth=2)
# Después lo mostraremos por pantalla
plt.show()
````

Se puede observar que aquí se adicionan elementos descriptivos: *etiquetas*, *leyendas* y *anotaciones*, que usualmente sirven para describir completamente una información que se proporciona a través de las gráficas.

## La función `plot`

La función `plot` está contenida en el módulo `pyplot` y básicamente con esta se produce cualquier gráfica bidimensional en coordenadas rectangulares. Esta función soporta varias maneras de ejecutarla dependiendo la cantidad de argumentos que se pasen.

La forma más básica de la función `plot` es pasarle un sólo argumento, como vimos en nuestro primer ejemplo.

Al pasarle un sólo argumento, este se toma como los valores de la coordenada vertical, y se asume que la horizontal varía de 0 a N-1, donde N es el número de elementos contenidos en la lista de valores que se introducen.

La sintaxis más habitual es introducir dos argumentos, como nuestro segundo ejemplo, donde el primero contiene una **lista X** que define los **valores de la coordenada horizonta**l, y el segundo una **lista Y** correspondiente a los **valores de la coordenada vertical**.

### Modificando el color, estilos y grosor de línea

La función **`plot`** acepta argumentos adicionales que sirven para modificar y controlar las características de la línea que se gráfica. Por ejemplo, se puede pasar un tercer argumento que contenga una combinación de color y estilo de línea:

````python
x = np.linspace(0, 30, 200)
y = np.exp(-0.1 * x) * np.cos(x)
plt.plot(x,y,color='red',linestyle='--')
plt.show()
````

El código anterior genera una gráfica con una línea en color rojo (`r`) y un estilo de línea discontinua (`--`).

Si en lugar del string **`--`** se coloca `go`, se obtiene una gráfica como la mostrada a continuación, podrá inferir que `g` refiere al color verde (green) y `o` justamente al uso de este caracter como símbolo para representar cada punto.

````python
x = np.linspace(0, 30, 100)
y = np.exp(-0.1 * x) * np.cos(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y, marker='o', markersize=4,color='green')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Diagrama de líneas — ejemplo')
plt.show()
````

En el siguiente [enlace](https://matplotlib.org/api/markers_api.html) se muestra una tabla con los símbolos (**markers**) disponibles para utilizar en la función `plot`.

En el siguiente [enlace](https://matplotlib.org/api/colors_api.html) puede consultar información respecto a los colores que puede abreviar mediante un sólo caracter.

Además de la forma anterior, también es posible especificar el color y estilo de línea utilizando ***keyword arguments***, por ejemplo:

````python
x = np.linspace(0, 30, 100)
y = np.exp(-0.1 * x) * np.cos(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y, linestyle='--', markersize=4,color='red')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Diagrama de líneas — ejemplo')
plt.show()
````

En ambos casos se especifica un cierto estilo de línea y color, con la diferencia
notoria de la sintaxis.

Utilizar *keyword arguments* es una manera más general, puesto que la definición con strings
no funciona para los casos en que se requieren colores que no se pueden especificar con
un sólo caracter, por ejemplo, Matplotlib dispone de un color llamado `coral` y
este no puede ser invocado mediante un sólo caracter, hace falta escribir todo el nombre.

````python
# Generar valores de X
x = np.linspace(0, 30, 100)

# Calcular la función seno amortiguada
y = np.sin(x) * np.exp(-x / 10)

# Crear gráfica
plt.plot(
    x,
    y,
    color='coral',
    marker='*',
    markersize=5,
    linestyle='-'
)

plt.show()
````

El grosor de línea se puede controlar mediante el *keyword argument* **`linewidth`**, por ejemplo;

````python
# Generar valores de X
x = np.linspace(0, 30, 100)

# Calcular la función seno amortiguada
y = np.sin(x) * np.exp(-x / 10)

# Crear gráfica
plt.plot(
    x,
    y,
    color='coral',
    marker='*',
    markersize=5,
    linestyle='-'
)

plt.show()
````

> La función `plot` soporta algunos *keyword arguments* acortados. Por ejemplo, se puede utilizar `ls` en lugar de `linestyle`, `lw` en lugar de `linewidth`, `ms` en lugar de `markersize`, `mfc` en lugar de `markerfacecolor`, entre otros.

## Título de gráfica, etiquetas de ejes y nombres de curvas

Por su naturaleza las gráficas nos sirven para presentar y/o visualizar información de ciertos datos, para lo cual se hace necesario especificar información descriptiva de lo que se muestra. Es muy común que se agreguen etiquetas a los ejes horizontal y vertical, así como el nombre de gráfica. Además, si se está graficando más de una curva, se hace necesario especificar a qué refiere cada una de ellas.

Por ejemplo, observe el siguiente código y la gráfica producida:

````python
# Dos curvas con leyenda
x = np.linspace(0, 30, 500)
y1 = np.exp(-0.1*x)*np.cos(x)
y2 = np.exp(-0.2*x)*np.sin(x)

plt.figure(figsize=(8,4))
plt.plot(x, y1, 'b-', label='Partícula 1 (exp(-0.1x) cos x)')
plt.plot(x, y2, 'r-', label='Partícula 2 (exp(-0.2x) sin x)')
plt.xlabel('Tiempo (s)')
plt.ylabel('Posición (mm)')
plt.title('Dos curvas en un mismo gráfico')
plt.legend()
plt.grid()
plt.show()
````

La instrucción **`xlabel`** coloca una etiqueta al eje horizontal, de manera similar **`ylabel`** lo hace para el eje
vertical. Con **`title`** adicionamos un título a la gráfica. La instrucción **`legend`** sirve para
colocar el recuadro con el *nombre* asignado a cada curva mediante el *keyword argument* **`label`**.

### Anotaciones

```python
plt.text(px, py, texto)
```

Donde **`px`** y **`py`** denotan las coordenadas en donde se colocará la anotación indicada en `texto`. Veamos un ejemplo:

````python
x = np.linspace(0, 30, 500)
y = np.exp(-0.1 * x) * np.cos(x)
plt.plot(x, y, color='purple')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.title('Gráfico de líneas')
plt.text(15, 0.5, 'Curva exponencial amortiguada')
plt.show()
````

### Gráfico de barras

Se utilizan para comparar datos entre diferentes categorías. Se pueden presentar horizontal o verticalmente. En este caso se utilizará la función **"bar()"**:

*   **`bar(x, y)`**: Dibuja un diagrama de barras verticales donde **`x`** es una lista con la posición de las barras en el eje X, e **`y`** es una lista con la altura de las barras en el eje Y.

Veamos como funciona con el siguiente ejemplo:


````python
# creamos las matrices con datos preparados
x1 = [0.25,1.25,2.25,3.25,4.25]
x2 = [0.75,1.75,2.75,3.75,4.75]
y1 = [10,30,75,5,92]
y2 = [5,50,20,35,50]

# configuramos los datos de cada matriz en el gráfico
plt.bar(x1, y1, label = "línea 1", color = "lightblue", width = 0.5)
plt.bar(x2, y2, label = "línea 2", color = "orange", width = 0.5)

# definimos el título y nombre de los ejes
plt.title("Gráfico de barras")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.legend()

plt.show()
````

### Histograma

Los histogramas son un tipo de gráfico de barras. **Se utilizan para mostrar una distribución de datos**. Son útiles cuando tienes matrices o una lista muy extensa de datos. Además, es una de las principales herramientas en el tratamiento de imágen.

En este caso, la función de visualización será **"hist()"**:

*   **`hist(x, bins)`**: Dibuja un histograma con las frecuencias resultantes de agrupar los datos de la lista **`x`** en las clases definidas por la lista **`bins`**.



````python
# Histograma
data = np.random.randint(0,100, size=1000)
sections = np.linspace(0,100,11)

plt.figure(figsize=(8,4))
plt.hist(data, sections, histtype='bar', rwidth=0.8)
plt.title('Histograma — ejemplo')
plt.xlabel('Tramos')
plt.ylabel('Frecuencia')
plt.show()
````

Aquí veréis otro ejemplo de los mismos datos con las barras más delgadas y con menos tramos:

````python
# Definimos unos tramos para los datos. Estos tramos serán 0, 25, 50...
sections = np.linspace(0, 100, 5)

# Configuramos los datos del gráfico
plt.hist(data, sections, histtype = "bar", rwidth = 0.1, color = "yellow")

plt.title("Histograma")
plt.xlabel("Tramos")
plt.ylabel("Valores/Tramo")
plt.show()
````

### Diagramas de sectores



El gráfico circular se divide en segmentos (o porciones de queso). Se utiliza para mostrar porcentajes o datos proporcionales en los que cada porción representa una categoría.

Matplotlib dispone de la función **"pie()"**, cuya sintaxis depende del grado de personalización y control que se requiera sobre la gráfica circular a dibujar.

Veamos un ejemplo, algo más complejo que en los casos anteriores. Vamos a suponer que se tienen los siguientes datos sobre algunos tipos de frutas, de los que se dispone de cierta cantidad:

| Nombre | Cantidad |
|-----|:----:|
| Apples | 35 |
| Bananas | 25 |
| Cherries | 25 |
| Dates | 15 |

Para representar el porcentaje del total del cual dispone cada uno, podemos trazar un diagrama de sectores. Para ello realizamos lo siguiente:

````python
# Ejemplo BÁSICO
array = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]

plt.pie(array, labels = mylabels)
plt.title("Gráfico circular")
plt.show() 
````

````python
# Ejemplo avanzado
datos = [7,8,2,2,5]
actividades = ["Dormir", "Trabajar", "Labores domésticas", "Estudiar", "Otros"]
colores = ["lightblue", "red", "orange", "green", "yellow"]

# En este caso mostramos un gráfico algo más completo:
#     - explode: remarcamos el caso que nos interese separándolo del gráfico
#     - shadow: muestra la sombra del gráfico
#     - autopct: muestra porcentajes automáticos en base a los datos introducidos
#     - startangle: hace que la gráfica empiece a mostrarse a partir de un ángulo concreto
plt.pie(datos, labels = actividades, colors = colores, startangle = 90,
        shadow = True, explode = (0.25,0,0,0,0), autopct = "%1.1f%%")

plt.title("Gráfico circular")
plt.show()
````

Personaliza los colores del gráfico utilizando una lista de color en formato hexadecimal o RGB.

````python
import matplotlib.pyplot as plt

# Datos
nombres = ['Ana', 'Juan', 'Diana', 'Catalina']
marcas = [20, 10, 25, 40]

# Colores personalizados en RGB
colores = [
    (1, 0.5, 0.5),
    (0.3, 0.8, 0.5),
    (0.6, 1, 0.4),
    (1, 0.8, 0.4)
]

# Crear gráfica
plt.pie(
    marcas,
    labels=nombres,
    autopct='%1.1f%%',
    colors=colores
)

plt.show()
````

Utiliza un **mapa de color** verde para personalizar los colores de los segmentos.

````python
import matplotlib.pyplot as plt
import numpy as np

# Datos
nombres = ['Ana', 'Juan', 'Diana', 'Catalina']
marcas = [20, 10, 25, 40]

# Generar colores utilizando un mapa de color
colores = plt.colormaps['Greens'](
    np.linspace(0.3, 1, len(marcas))
)

# Crear gráfica
plt.pie(
    marcas,
    labels=nombres,
    autopct='%1.1f%%',
    colors=colores
)

plt.show()
````

Separe el segmento correspondiente a Diana y aplique un estilo con sombra al gráfico:



````python
import matplotlib.pyplot as plt
import numpy as np

# Datos
nombres = ['Ana', 'Juan', 'Diana', 'Catalina']
marcas = [20, 10, 25, 40]

# Mapa de colores verdes
colores = plt.colormaps['Greens'](
    np.linspace(0.3, 1, len(marcas))
)

# Separar el segmento de Diana
explode = (0, 0, 0.1, 0)

# Crear gráfica
plt.pie(
    marcas,
    labels=nombres,
    autopct='%1.1f%%',
    colors=colores,
    explode=explode,
    shadow=True
)

plt.show()
````
